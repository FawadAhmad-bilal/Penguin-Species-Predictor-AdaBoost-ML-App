```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df=sns.load_dataset("penguins")
```

```python
df=df.dropna()
```

```python
df.isnull().sum()
```

```python
for i in df.columns:
    print(df[i].unique())
    print(df[i].value_counts())
    print("="*30)
```

```python
df.info()
```

```python
df["sex"] = df["sex"].map({"Male":0,"Female":1})
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
df["species"]=le.fit_transform(df["species"])
df["island"]=le.fit_transform(df["island"])
```

```python
df.head(25)
```

```python
df["island"].unique()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,mean_squared_error,confusion_matrix,classification_report,ConfusionMatrixDisplay
```

```python
x=df.drop("species",axis=1)
y=df["species"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dt=DecisionTreeClassifier(max_depth=1)
```

```python
AdaBoostClassifier().get_params()
```

```python
dt.fit(x_train,y_train)
```

```python
pre_y=dt.predict(x_test)
```

```python
accuracy_score(y_test,pre_y)
```

```python
adb=AdaBoostClassifier(
    estimator=dt,
    n_estimators=10,
    learning_rate=1,
    random_state=42)
```

```python
adb.fit(x_train,y_train)
```

```python
re_y=adb.predict(x_test)
```

```python
accuracy_score(y_test,re_y)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
cm=confusion_matrix(y_test,re_y)
```

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=adb.classes_)
disp.plot(cmap='Blues')
plt.title('AdaBoost Confusion Matrix')
plt.show()
```

```python
importances = adb.feature_importances_
features = x_train.columns
# Plot
plt.figure(figsize=(10, 6))
plt.barh(features, importances, color='steelblue')
plt.xlabel('Importance')
plt.title('AdaBoost Feature Importances')
plt.tight_layout()
plt.show()
```
